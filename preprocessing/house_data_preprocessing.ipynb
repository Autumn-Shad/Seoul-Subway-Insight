```python
import pandas as pd 
```

```python
import os
import glob

csv_files = glob.glob('./*.csv')

# 각 파일을 반복하며 읽어와 하나의 데이터프레임으로 합칩니다.
all_data = pd.DataFrame()
for f in csv_files:
    try:
        df = pd.read_csv(f, encoding='utf-8')
    except UnicodeDecodeError:
        # If cp949 fails, try euc-kr:
        try:
            df = pd.read_csv(f, encoding='cp949')
        except UnicodeDecodeError:
            # As a last resort, try 'latin1' (though data might look wrong if it's Korean)
            df = pd.read_csv(f, encoding='latin1')

all_data = pd.concat([all_data, df], ignore_index=True)
```

```python
## 역 이름 뒤에 -역 제외하기
station_name=[]

for i in all_data['NRB_SWST_NM']:
    if '역' not in i:
        station_name.append(i)
    else:
        if i[-1] == '역':
            station_name.append(i[:-1])
            continue
        if '(' in i:
            if '호선' in i:
                station_name.append(i.split('(')[0][:-1])
                continue
            else:
                if i.split('(')[0][-1] == '역':
                    station_name.append(i.split('(')[0][:-1]+'('+i.split('(')[1][:])
                    continue
                elif len(i.split(')')[1])>0 and i.split(')')[1][0] == '역':
                    station_name.append(i.split(')')[0][:]+')')
                    continue
        if ' ' in i:
            if i.split(' ')[0][-1] == '역':
                station_name.append(i.split(' ')[0][:-1]+i.split(' ')[1][:])
                continue
        else:
            station_name.append(i)

print(len(station_name))
print(all_data.shape)
```

```python
all_data['NRB_SWST_NM']=station_name
```
