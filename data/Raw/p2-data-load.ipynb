# CREATE TABLE 테이블 생성

# 데이터 로드
- s3에서 폴더별로 구분하여 로딩

```
COPY INTO autumn_gs_seoulmetro.raw_data.metro_station_info
FROM 's3://autumn-gs-metro/metro_raw_data/seoul_metro_station_info.csv'
credentials = (
    AWS_KEY_ID=$AWS_ID 
    AWS_SECRET_KEY=$AWS_KEY
)
FILE_FORMAT = (
    TYPE='CSV' 
    SKIP_HEADER=1 
    FIELD_OPTIONALLY_ENCLOSED_BY='"'
);
```

```
COPY INTO autumn_gs_seoulmetro.raw_data.metro_daily_traffic
FROM 's3://autumn-gs-metro/metro_raw_data/seoul_metro_daily/' 
credentials = (
    AWS_KEY_ID=$AWS_ID 
    AWS_SECRET_KEY=$AWS_KEY
)
FILE_FORMAT = (
    TYPE='CSV' 
    SKIP_HEADER=1 
    FIELD_OPTIONALLY_ENCLOSED_BY='"'
);
```
